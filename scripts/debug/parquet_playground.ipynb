# Parquet Playground (Doom dataset)

Use this notebook to inspect one parquet shard and decode byte fields.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import pyarrow.parquet as pq
```

```python
DATA_DIR = Path('/mnt/weka/home/hao.zhang/mhuo/traindata_0204_1600/preprocessed')
parquet_files = sorted(DATA_DIR.rglob('*.parquet'))
print(f'Found {len(parquet_files)} parquet files under {DATA_DIR}')
print('First 5 files:')
for p in parquet_files[:5]:
    print(' -', p)
```

```python
# Pick one file to inspect
file_idx = 0
parquet_path = parquet_files[file_idx]

pf = pq.ParquetFile(parquet_path)
print('FILE:', parquet_path)
print('ROWS:', pf.metadata.num_rows)
print('ROW_GROUPS:', pf.num_row_groups)
print('SCHEMA:')
print(pf.schema)

df = pq.read_table(parquet_path).to_pandas()
with pd.option_context('display.max_columns', 50, 'display.width', 200):
    display(df.head(3))
```

```python
def decode_blob(row, prefix):
    b = row[f'{prefix}_bytes']
    shape = row[f'{prefix}_shape']
    dtype = row[f'{prefix}_dtype']
    if b is None or len(b) == 0:
        return None
    arr = np.frombuffer(b, dtype=np.dtype(dtype))
    return arr.reshape(shape)

row0 = df.iloc[0]
for k in ['vae_latent', 'clip_feature', 'first_frame_latent', 'mouse_cond', 'keyboard_cond']:
    arr = decode_blob(row0, k)
    if arr is None:
        print(f'{k}: empty')
    else:
        print(f'{k}: shape={arr.shape}, dtype={arr.dtype}, min={arr.min():.4f}, max={arr.max():.4f}')

mouse = decode_blob(row0, 'mouse_cond')
keyboard = decode_blob(row0, 'keyboard_cond')

if mouse is not None:
    print('\nmouse_cond (first 12 rows):')
    print(np.array2string(mouse[:12], precision=3, suppress_small=True))
    print('mouse unique values:', np.unique(np.round(mouse, 3)))
    display(pd.DataFrame(mouse, columns=['mouse_dx', 'mouse_dy']).head(20))

if keyboard is not None:
    print('\nkeyboard_cond (first 12 rows):')
    print(np.array2string(keyboard[:12], precision=3, suppress_small=True))
    print('keyboard unique values:', np.unique(np.round(keyboard, 3)))
    key_cols = ['key_w', 'key_a', 'key_s', 'key_d', 'mouse_l', 'mouse_r']
    if keyboard.shape[1] == len(key_cols):
        display(pd.DataFrame(keyboard, columns=key_cols).head(20))
    else:
        display(pd.DataFrame(keyboard).head(20))
```
